```python
import json
messages = [
    "There is a major water leak in Building C on Level 2. Water is still flowing and is close to electrical equipment. Please send someone urgently.",

    "I want to report that one of our suppliers may not meet our sustainability procurement policy. I do not have evidence of an immediate safety issue, but the contract is due for renewal next month and I think someone should review it.",

    "An employee named Priya Shah has a medical condition and says the accessible entrance near the main building has been blocked for two days. She cannot safely use the stairs and needs assistance accessing the building."
]

def classify_message(message):
    text = message.lower()

    # Issue category
    if any(word in text for word in [
        "supplier", "procurement", "policy", "compliance", "contract"
    ]):
        issue_category = "Governance"

    elif any(word in text for word in [
        "accessible", "accessibility", "employee", "medical",
        "wellbeing", "workplace"
    ]):
        issue_category = "Social"

    elif any(word in text for word in [
        "water", "leak", "recycling", "waste",
        "energy", "electricity", "emission"
    ]):
        issue_category = "Environmental"

    else:
        issue_category = "Other"

    # Urgency
    if any(word in text for word in [
        "electrical equipment", "immediate danger",
        "life-threatening", "fire", "severe injury", "urgently"
    ]):
        urgency = "Critical"

    elif any(word in text for word in [
        "blocked for two days", "cannot safely",
        "can't safely", "still happening", "medical condition"
    ]):
        urgency = "High"

    elif any(word in text for word in [
        "renewal next month", "review it", "policy", "contract"
    ]):
        urgency = "Medium"

    else:
        urgency = "Low"

    # Sentiment
    if any(word in text for word in [
        "major", "blocked", "not meet", "concern",
        "leak", "cannot safely", "can't safely", "urgent"
    ]):
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    # Follow-up
    follow_up_required = (
        "Yes"
        if urgency != "Low"
        or any(word in text for word in ["report", "review", "blocked", "leak"])
        else "No"
    )

    # Recommended team
    if issue_category == "Governance":
        recommended_team = "Compliance"

    elif issue_category == "Social":
        recommended_team = (
            "Facilities"
            if any(word in text for word in [
                "accessible", "entrance", "stairs", "building"
            ])
            else "HR"
        )

    elif issue_category == "Environmental":
        recommended_team = (
            "Facilities"
            if any(word in text for word in [
                "leak", "water", "recycling", "energy", "electricity"
            ])
            else "Sustainability"
        )

    else:
        recommended_team = "Other"

    # Data sensitivity
    data_sensitivity_risk = (
        "High"
        if any(word in text for word in [
            "medical", "employee named", "named"
        ])
        else "Low"
    )

    return {
        "issue_category": issue_category,
        "urgency": urgency,
        "sentiment": sentiment,
        "follow_up_required": follow_up_required,
        "recommended_team": recommended_team,
        "data_sensitivity_risk": data_sensitivity_risk
    }


for i, message in enumerate(messages, 1):
    print(f"\nTEST {i}")
    print("Message:", message)
    print("Rule-based result:")
    print(json.dumps(classify_message(message), indent=2))
```
